```python
"""Calculator module for performing arithmetic operations."""

from __future__ import annotations
from typing import Any


class Calculator:
    """A simple calculator class supporting basic arithmetic operations."""

    def __init__(self) -> None:
        """Initialize the calculator with a zero result."""
        self._result: float = 0.0

    def add(self, a: float, b: float) -> float:
        """Add two numbers together.

        Args:
            a: The first number.
            b: The second number.

        Returns:
            The sum of a and b.

        Raises:
            ValueError: If either argument is not a number.
        """
        if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
            raise ValueError("Both arguments must be numbers.")
        return a + b

    def subtract(self, a: float, b: float) -> float:
        """Subtract b from a.

        Args:
            a: The first number.
            b: The second number.

        Returns:
            The result of a - b.

        Raises:
            ValueError: If either argument is not a number.
        """
        if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
            raise ValueError("Both arguments must be numbers.")
        return a - b

    def multiply(self, a: float, b: float) -> float:
        """Multiply two numbers together.

        Args:
            a: The first number.
            b: The second number.

        Returns:
            The result of a * b.

        Raises:
            ValueError: If either argument is not a number.
        """
        if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
            raise ValueError("Both arguments must be numbers.")
        return a * b

    def divide(self, a: float, b: float) -> float:
        """Divide a by b.

        Args:
            a: The dividend.
            b: The divisor.

        Returns:
            The result of a / b.

        Raises:
            ValueError: If b is zero or either argument is not a number.
        """
        if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
            raise ValueError("Both arguments must be numbers.")
        if b == 0:
            raise ValueError("Division by zero is not allowed.")
        return a / b

    def __add__(self, other: float) -> float:
        """Add another number to self.

        Args:
            other: The number to add.

        Returns:
            self + other.

        Raises:
            ValueError: If other is not a number.
        """
        if not isinstance(other, (int, float)):
            raise ValueError("Other must be a number.")
        return self.add(other)

    def __radd__(self, other: float) -> float:
        """Add other to self (reverse order).

        Args:
            other: The number to add.

        Returns:
            other + self.

        Raises:
            ValueError: If other is not a number.
        """
        if not isinstance(other, (int, float)):
            raise ValueError("Other must be a number.")
        return self.add(other)

    def __sub__(self, other: float) -> float:
        """Subtract other from self.

        Args:
            other: The number to subtract.

        Returns:
            self - other.

        Raises:
            ValueError: If other is not a number.
        """
        if not isinstance(other, (int, float)):
            raise ValueError("Other must be a number.")
        return self.subtract(other)

    def __rsub__(self, other: float) -> float:
        """Subtract other from self (reverse order).

        Args:
            other: The number to subtract.

        Returns:
            self - other.

        Raises:
            ValueError: If other is not a number.
        """
        if not isinstance(other, (int, float)):
            raise ValueError("Other must be a number.")
        return self.subtract(other)

    def __mul__(self, other: float) -> float:
        """Multiply self by other.

        Args:
            other: The number to multiply by.

        Returns:
            self * other.

        Raises:
            ValueError: If other is not a number.
        """
        if not isinstance(other, (int, float)):
            raise ValueError("Other must be a number.")
        return self.multiply(other)

    def __rmul__(self, other: float) -> float:
        """Multiply other by self (reverse order).

        Args:
            other: The number to multiply by.

        Returns:
            other * self.

        Raises:
            ValueError: If other is not a number.
        """
        if not isinstance(other, (int, float)):
            raise ValueError("Other must be a number.")
        return self.multiply(other)

    def __truediv__(self, other: float) -> float:
        """Divide self by other.

        Args:
            other: The number to divide by.

        Returns:
            self / other.

        Raises:
            ValueError: If other is zero or either argument is not a number.
        """
        if not isinstance(other, (int, float)):
            raise ValueError("Other must be a number.")
        if other == 0:
            raise ValueError("Division by zero is not allowed.")
        return self.divide(other)

    def __rtruediv__(self, other: float) -> float:
        """Divide other by self (reverse order).

        Args:
            other: The number to divide by.

        Returns:
            other / self.

        Raises:
            ValueError: If self is zero or either argument is not a number.
        """
        if not isinstance(other, (int, float)):
            raise ValueError("Other must be a number.")
        if self == 0:
            raise ValueError("Division by zero is not allowed.")
        return self.divide(other)

    def __repr__(self) -> str:
        """Return a string representation of the calculator.

        Returns:
            A string like "Calculator(result=12.5)".
        """
        return f"Calculator(result={self._result})"

    def __str__(self) -> str:
        """Return a string representation of the calculator.

        Returns:
            A string like "Calculator(result=12.5)".
        """
        return self.__repr__()

    def __eq__(self, other: object) -> bool:
        """Compare two calculators for equality.

        Args:
            other: Another calculator instance.

        Returns:
            True if both calculators have the same result.
        """
        if not isinstance(other, Calculator):
            return NotImplemented
        return self._result == other._result

    def __hash__(self) -> int:
        """Return a hash value for the calculator.

        Returns:
            A hash value.
        """
        return hash((self._result,))


# --- Demo ---
if __name__ == "__main__":
    calculator = Calculator()

    # Basic operations
    print(f"3 + 5 = {calculator.add(3.2, 5)}")
    print(f"10 - 4 = {calculator.subtract(10, 4)}")
    print(f"7 * 8 = {calculator.multiply(7, 8)}")
    print(f"100 / 4 = {calculator.divide(100, 4)}")

   
    # Edge cases
    print(f"-3 + 5 = {calculator.add(-3, 5)}")
    print(f"1000 / -1 = {calculator.divide(1000, -1)}")
    #print(f"1000 / 0 = {calculator.divide(1000, 0)}")  # Should raise ValueError

    # Non-numeric inputs
    try:
        calculator.add("hello", 5)
    except ValueError as e:
        print(f"Error: {e}")

    # Equality
    print(f"Calculator(12.5) == Calculator(12.5): {calculator == Calculator(12.5)}")

    # Test with negative numbers
    print(f"-3 - (-5) = {calculator.subtract(-3, -5)}")
    print(f"(-3) * (-5) = {calculator.multiply(-3, -5)}")
    print(f"100 / -4 = {calculator.divide(100, -4)}")
```
