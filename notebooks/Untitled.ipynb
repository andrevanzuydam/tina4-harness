```python
"""
A complete demonstration of how classes work in Python.
This module covers: inheritance, properties, decorators,
methods with parameters/defaults, multiple inheritance, and more.
"""


# ============================================================
# 1. BASIC CLASS DEFINITION AND INSTANCE CREATION
# ============================================================

class Animal:
    """Base class for all animals."""

    def __init__(self):
        self.name = ""          # ← Instance variable (default value)
        self.color = "gray"     # ← Class attribute shared by ALL instances


class Dog(Animal):              # ← Inheritance from base class Animal
    def bark(self, name: str = "") -> None:  # ← Method with default parameter
        print(f"{self.name} says Woof!")

    @property
    def breed(self) -> str:   # ← Property (getter), returns a string
        return "Dog"


class Cat(Animal):            # ← Another subclass inheriting from Animal
    def meow(self, name: str = "") -> None:  # ← Method with default parameter
        print(f"{self.name} says Meow!")

    @property
    def breed(self) -> str:   # ← Property (getter), returns a string
        return "Cat"


class Bird(Animal):            # ← Third subclass inheriting from Animal
    def chirp(self, name: str = "") -> None:  # ← Method with default parameter
        print(f"{self.name} says Chirp!")

    @property
    def breed(self) -> str:   # ← Property (getter), returns a string
        return "Bird"


# ============================================================
# 2. MULTIPLE INHERITANCE
# ============================================================

class Vehicle:
    """Base class for all vehicles."""

    def __init__(self):
        self.brand = ""          # ← Class attribute shared by ALL instances
        self.year = None         # ← Instance variable (default value)


class Car(Vehicle):            # ← Inherits from Vehicle
    def drive(self, speed: int = 0) -> str:   # ← Method with default parameter
        return f"{self.brand} is driving at {speed} km/h"

    @property
    def fuel_type(self) -> str:      # ← Property (getter), returns a string
        return "Gasoline"


class Truck(Vehicle):              # ← Inherits from Vehicle AND Car
    def drive(self, speed: int = 0) -> None:   # ← Method with default parameter
        print(f"{self.brand} is driving at {speed} km/h")

    @property
    def fuel_type(self) -> str:      # ← Property (getter), returns a string
        return "Diesel"


class Motorcycle(Vehicle):          # ← Inherits from Vehicle AND Car, but NOT Truck
    def ride(self, speed: int = 0) -> None:   # ← Method with default parameter
        print(f"{self.brand} is riding at {speed} km/h")
```
