```python
# Right-click me → "Harness: Check Code".  Or select the loop → "Harness: Fix Code".
def totals(items):
    t=0
    for i in range(len(items)):
        t=t+items[i]['price']*items[i]['qty']
    return t

totals([{'price':2,'qty':3},{'price':5,'qty':1}])
```

```python
"""
A simple calculator that supports:
- Addition, subtraction, multiplication, division
- Parentheses and parentheses
- Negative numbers
- Floating-point numbers
"""

from calculator import Calculator


def main() -> None:
    """Demonstrate the calculator."""
    calc = Calculator()

    print("=== Calculator ===")
    print(f"2 + 3 = {calc.evaluate('2 + 3')}")
    print(f"(2 + 3) * 4 = {calc.evaluate('(2 + 3) * 4')}")
    print(f"- 5 + 3 = {calc.evaluate('- 5 + 3')}")
    print(f"10 / 2 = {calc.evaluate('10 / 2')}")
    print(f"(10 / 2) + 3 = {calc.evaluate('(10 / 2) + 3')}")

    # Custom function
    def square(x: float) -> float:
        """Square a number."""
        return x * x

    calc = Calculator()
    print(f"square(5) = {calc.evaluate('square(5)')}")

    # Evaluate a complex expression
    print(f"2 * 3 + 4 / (5 - 1) = {calc.evaluate('2 * 3 + 4 / (5 - 1)')}")

    # Evaluate a more complex expression with parentheses
    print(f"(2 + 3) * (4 - 1) = {calc.evaluate('(2 + 3) * (4 - 1)')}")

    # Evaluate with floating-point numbers
    print(f"10.5 / 2.5 = {calc.evaluate('10.5 / 2.5')}")
    print(f"(10.5 / 2.5) * 4 = {calc.evaluate('(10.5 / 2.5) * 4')}")


if __name__ == "__main__":
    main()
```
