```python
# Right-click me → "Harness: Check Code".  Or select the loop → "Harness: Fix Code".
def totals(items):
    t=0
    for i in range(len(items)):
        t=t+items[i]['price']*items[i]['qty']
    return t

totals([{'price':2,'qty':3},{'price':5,'qty':1}])
```

### Add your own tool, then run the agent

```python
@tool
def word_count(text: str) -> str:
    "Count the words in some text. args: text"
    return str(len(text.split()))

run_agent("Use word_count on 'the quick brown fox jumps' and give the final answer.")
```
